4. Найти среднее значение строки матрицы и вычесть его из каждого элемента строки

````python
z = np.random.rand(5, 10)
print("Исходная матрица:", z)
row_means = z.mean(axis=1)
print("\nСредние значения строк:", row_means)
mat = z - row_means[:, np.newaxis]
print("\nМатрица после вычитания:\n", mat)
````

6. Дан массив 10x2 представляющий собой точки в декартовой системе координат, преобразовать в полярную

````python
import numpy as np
z = np.random.random((10, 2))
X, Y = z[:, 0], z[:, 1]
R = np.hypot(X, Y)
T = np.arctan2(Y, X)
polar = np.column_stack((R, T))

print("Декартовы координаты:\n", z)
print("\nПолярные координаты:\n", polar)
````
