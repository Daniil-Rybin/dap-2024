1. Создать массив 10x10 со случайными значениями, найти минимум и максимум (низкая)

````python
array = np.random.randint(-10, 10, size = (10,10))
print(f"Массив: {array}")
min = array.min()
max = array.max()
print(f"Минимум: {min}")
print(f"Максимум: {max}")
````

2. Создать 8x8 матрицу и заполнить её единицами и нулями в шахматном порядке

````python
matrix = np.ones((8, 8), dtype=int)
matrix[1::2, ::2] = 0
matrix[::2, 1::2] = 0
print(matrix)
````

3. Дан массив из 11 элементов, поменять знак у элементов, значения которых между 3 и 8

````python
z = np.arange(11)
z[(z > 3) & (z < 8)] *= -1
print(z)
````

4. Найти среднее значение строки матрицы и вычесть его из каждого элемента строки

````python
z = np.random.rand(5, 10)
print("Исходная матрица:", z)
row_means = z.mean(axis=1)
print("\nСредние значения строк:", row_means)
mat = z - row_means[:, np.newaxis]
print("\nМатрица после вычитания:\n", mat)
````

5. Поменяйте 1 и 2 строки матрицы

````python
z = np.arange(25).reshape(5, 5)
print("До:\n", z)
z[[0, 1]] = z[[1, 0]]
print("\nПосле:\n", z)
````

6. Дан массив 10x2 представляющий собой точки в декартовой системе координат, преобразовать в полярную

````python
import numpy as np
z = np.random.random((10, 2))
X, Y = z[:, 0], z[:, 1]
R = np.hypot(X, Y)
T = np.arctan2(Y, X)
polar = np.column_stack((R, T))

print("Декартовы координаты:\n", z)
print("\nПолярные координаты:\n", polar)
````

7. Заменить максимальный элемент массива на ноль

````python
z = np.random.random(10)
print("Исходный массив:\n", z)
print("\nМаксимальный элемент: ", z.max())
z[z.argmax()] = 0
print("\nПосле замены:\n", z)
````

8. Дан массив, добавить 1 к каждому элементу с индексом, заданным в другом массиве. Добавление производится только 1 раз

````python
z = np.ones(10)
i = np.random.randint(0, len(z), 20)
print("Индексы:\n", i)
z[np.unique(i)] += 1
print("\nИтоговый массив:\n", z)
````

9. Дан вектор [1, 2, 3, 4, 5], построить новый вектор с тремя нулями между каждым значением

````python
z = np.array([1, 2, 3, 4, 5])
k = 3
new_len = len(z) + (len(z) - 1) * k
result = np.zeros(new_len, dtype=z.dtype)
result[::k + 1] = z
print(result)
````

10. Найти ближайшее к заданному значению число в заданном массиве

````python
z = np.arange(100)
v = np.random.uniform(0, 100)
index = np.abs(z - v).argmin()
closest_value = z[index]
print(f"Искомое значение: {v:.4f}")
print(f"Ближайший элемент: {closest_value} (на индексе {index})")
````

**10**.  В чем преимущество массивов NumPy по сравнению с обычными списками.  Чем обусловлено это преимущество?

**Преимущества**: быстрее в сотни раз, занимают меньше памяти.

**Причины**:

Непрерывная память - данные кэшируются процессором), не в виде разрозненных указателей.

Один тип данных (dtype) - нет расходов на динамическую типизацию и проверку типов на каждом шаге.
