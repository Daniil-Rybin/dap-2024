4. Сколько мужчин / женщин находилось на борту?

4. Выведите распределение переменной Pclass (социально-экономический статус) и это же распределение, только для мужчин / женщин по отдельности. Сколько было мужчин 2-го класса?

````python
print("\n---Распределение Pclass (общее)---")
print(df['Pclass'].value_counts().sort_index())

print("\n---Распределение Pclass по полу---")
print(df.groupby(['Sex', 'Pclass']).size().unstack(fill_value=0))

men_2nd_class = df[(df['Sex'] == 'male') & (df['Pclass'] == 2)].shape[0]
print(f"\nМужчин 2-го класса: {men_2nd_class}")
````

5. Каковы медиана и среднеквадрати́чное отклонение платежей (Fare)? Округлите до 2 десятичных знаков.

````python
median_fare = df['Fare'].median()
std_fare = df['Fare'].std()

print(f"Медиана: {round(median_fare, 2)}")
print(f"Стандартное отклонение: {round(std_fare, 2)}")
````

6. Правда ли, что люди моложе 30 лет выживали чаще, чем люди старше 60 лет? Каковы доли выживших в обеих группах?

````python
young = df[df['Age'] < 30]
old = df[df['Age'] > 60]

young_survival_rate = young['Survived'].mean()
old_survival_rate = old['Survived'].mean()

print(f"Доля выживших среди людей моложе 30 лет: {round(young_survival_rate, 2)}")
print(f"Доля выживших среди людей старше 60 лет: {round(old_survival_rate, 2)}")
print(f"Люди моложе 30 лет выживали чаще: {young_survival_rate > old_survival_rate}")
````

7. Правда ли, что женщины выживали чаще мужчин? Каковы доли выживших в обеих группах?

````python
female_survival_rate = df[df['Sex'] == 'female']['Survived'].mean()
male_survival_rate = df[df['Sex'] == 'male']['Survived'].mean()

print(f"Доля выживших среди женщин: {round(female_survival_rate, 2)}")
print(f"Доля выживших среди мужчин: {round(male_survival_rate, 2)}")
print(f"Женщины выживали чаще мужчин: {female_survival_rate > male_survival_rate}")
````

8.  Найдите самое популярное имя среди пассажиров Титаника мужского пола?

````python
df['FirstName'] = df['Name'].str.extract(r'\.\s+([A-Za-z]+)')

male_names = df[df['Sex'] == 'male']['FirstName']

most_popular_male_name = male_names.value_counts().idxmax()
print(f"Самое популярное имя среди мужчин: {most_popular_male_name}")
````



9. Как отличается средний возраст мужчин / женщин в зависимости от класса обслуживания? Выберите верные утверждения:

    1.В среднем мужчины 1-го класса старше 40 лет

    2.В среднем женщины 1-го класса старше 40 лет

    3.Мужчины всех классов в среднем старше женщин того же класса
    
    4.В среднем люди в 1 классе старше, чем во 2-ом, а те старше представителей 3-го класса



````python
mean_age = df.groupby(['Sex', 'Pclass'])['Age'].mean().round(2)
print("---Средний возраст по полу и классу---")
print(mean_age)
mean_age_by_class = df.groupby('Pclass')['Age'].mean().round(2)
print("\n---Средний возраст по классу---")
print(mean_age_by_class)

print("\n---Утверждения (в среднем)---")
print(f"1) Мужчины 1-го класса старше 40 лет: {mean_age[('male', 1)] > 40}")
print(f"2) Женщины 1-го класса старше 40 лет: {mean_age[('female', 1)] > 40}")
print(f"3) Мужчины всех классов старше женщин того же класса:")
for pclass in [1, 2, 3]:
    print(f"   Класс {pclass}: мужчины {mean_age[('male', pclass)]} vs женщины {mean_age[('female', pclass)]} -> {mean_age[('male', pclass)] > mean_age[('female', pclass)]}")
print(f"4) Люди 1 класса старше, чем во 2-ом, а те старше 3-го: {mean_age_by_class[1] > mean_age_by_class[2] > mean_age_by_class[3]}")
````

10. Сравните возраст у спасенных и у погибших пасажиров. Средний возраст погибших выше, верно?

````python
mean_age_survived = df[df['Survived'] == 1]['Age'].mean()
mean_age_died = df[df['Survived'] == 0]['Age'].mean()

print(f"Средний возраст спасённых: {round(mean_age_survived, 2)}")
print(f"Средний возраст погибших:  {round(mean_age_died, 2)}")
print(f"Средний возраст погибших выше: {mean_age_died > mean_age_survived}")
````

### Ответы на контрольные вопросы

**1. Что такое медиана? Рассчитать медиану для предложенного ряда чисел.**

Медиана - это значение, которое делит упорядоченный ряд чисел пополам: половина значений меньше медианы, половина - больше. Если количество элементов чётное, медиана - среднее арифметическое двух центральных значений.

**2. Что такое дисперсия? Рассчитать дисперсию для предложенного ряда чисел.**

Дисперсия - мера разброса значений относительно среднего. Это среднее квадратов отклонений от среднего.

**3. Что такое среднеквадратичное отклонение? Рассчитать среднеквадратичное отклонение для предложенного ряда чисел.**

Среднеквадратичное (стандартное) отклонение - корень из дисперсии. Показывает, насколько в среднем значения отклоняются от среднего.

**4. Что такое Series и чем он отличается от DataFrame?**

Series — одномерный маркированный массив (аналог столбца в таблице). Состоит из индекса и значений.

DataFrame — двумерная таблица (аналог листа Excel). Состоит из индекса строк, названий столбцов и значений.
